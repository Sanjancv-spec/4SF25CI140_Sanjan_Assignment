```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

print("=" * 60)
print("1. DataFrame with 3 columns, 5 rows & product column")
print("=" * 60)
df1 = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)), columns=['Col_A', 'Col_B', 'Col_C'])
df1['Product_AB'] = df1['Col_A'] * df1['Col_B']
print(df1)

print("\n" + "=" * 60)
print("2. DataFrame with 3 columns, 4 rows: Row-wise & Column-wise Sum")
print("=" * 60)
df2 = pd.DataFrame(np.random.randint(1, 50, size=(4, 3)), columns=['A', 'B', 'C'])
print("Original DataFrame:\n", df2)
row_sum = df2.sum(axis=1)
col_sum = df2.sum(axis=0)
print("\nRow-wise Sum:\n", row_sum)
print("\nColumn-wise Sum:\n", col_sum)

print("\n" + "=" * 60)
print("3. DataFrame with NaN values: Impute with Column Mean")
print("=" * 60)
df3 = pd.DataFrame(np.random.randint(10, 100, size=(5, 3)).astype(float), columns=['X', 'Y', 'Z'])
df3.iloc[0, 1] = np.nan
df3.iloc[2, 0] = np.nan
df3.iloc[3, 2] = np.nan
print("DataFrame before Imputation:\n", df3)
df3_imputed = df3.fillna(df3.mean())
print("\nDataFrame after Mean Imputation:\n", df3_imputed)

print("\n" + "=" * 60)
print("4. DataFrame with 4 columns, 6 rows: Drop rows with NaN")
print("=" * 60)
df4 = pd.DataFrame(np.random.randint(1, 50, size=(6, 4)).astype(float), columns=['W', 'X', 'Y', 'Z'])
df4.iloc[1, 2] = np.nan
df4.iloc[4, 0] = np.nan
print("DataFrame with NaNs:\n", df4)
df4_dropped = df4.dropna()
print("\nDataFrame after dropping NaNs:\n", df4_dropped)

print("\n" + "=" * 60)
print("5. Data Aggregation: Category & Value (Sum and Mean)")
print("=" * 60)
categories_5 = np.random.choice(['A', 'B', 'C'], size=10)
values_5 = np.random.randint(10, 100, size=10)
df5 = pd.DataFrame({'Category': categories_5, 'Value': values_5})
print("Original DataFrame:\n", df5)
agg5 = df5.groupby('Category')['Value'].agg(['sum', 'mean']).reset_index()
print("\nSum and Mean by Category:\n", agg5)

print("\n" + "=" * 60)
print("6. Product, Category, and Sales: Total Sales by Category")
print("=" * 60)
df6 = pd.DataFrame({
    'Product': [f'Product_{i}' for i in range(1, 10)],
    'Category': np.random.choice(['Electronics', 'Clothing', 'Groceries'], size=9),
    'Sales': np.random.randint(100, 1000, size=9)
})
print("Sales Data:\n", df6)
total_sales = df6.groupby('Category')['Sales'].sum().reset_index()
print("\nTotal Sales per Category:\n", total_sales)

print("\n" + "=" * 60)
print("7. Merging Two DataFrames on a Common Column")
print("=" * 60)
df7_employees = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David']
})
df7_departments = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104],
    'Department': ['AI & ML', 'Data Engineering', 'Cybersecurity', 'Cloud Computing']
})
merged_df = pd.merge(df7_employees, df7_departments, on='Emp_ID')
print("Merged DataFrame:\n", merged_df)

print("\n" + "=" * 60)
print("8. MultiIndex DataFrame: Indexing and Slicing")
print("=" * 60)
multi_levels_8 = [
    ['Dept_A', 'Dept_A', 'Dept_B', 'Dept_B'],
    ['Team_1', 'Team_2', 'Team_1', 'Team_2']
]
index_8 = pd.MultiIndex.from_arrays(multi_levels_8, names=('Department', 'Team'))
df8 = pd.DataFrame(np.random.randint(50, 100, size=(4, 2)), index=index_8, columns=['Metric_1', 'Metric_2'])
print("MultiIndex DataFrame:\n", df8)
print("\nIndexing 'Dept_A':\n", df8.loc['Dept_A'])
print("\nSlicing specific cell ('Dept_A', 'Team_1'):\n", df8.loc[('Dept_A', 'Team_1')])

print("\n" + "=" * 60)
print("9. MultiIndex with Category & SubCategory: Sum of Values")
print("=" * 60)
cat_9 = ['Hardware', 'Hardware', 'Furniture', 'Furniture']
subcat_9 = ['Laptops', 'Monitors', 'Chairs', 'Desks']
multi_idx_9 = pd.MultiIndex.from_arrays([cat_9, subcat_9], names=['Category', 'SubCategory'])
df9 = pd.DataFrame({'Sales_Units': np.random.randint(20, 200, size=4)}, index=multi_idx_9)
print("MultiIndex Sales Data:\n", df9)
sum_by_category = df9.groupby(level='Category').sum()
sum_by_subcat = df9.groupby(level=['Category', 'SubCategory']).sum()
print("\nSum by Category:\n", sum_by_category)
print("\nSum by Category & SubCategory:\n", sum_by_subcat)

print("\n" + "=" * 60)
print("10. Applying Functions: Double Values in a 3x5 DataFrame")
print("=" * 60)
df10 = pd.DataFrame(np.random.randint(1, 25, size=(5, 3)), columns=['Col1', 'Col2', 'Col3'])
print("Original DataFrame:\n", df10)
df10_doubled = df10.map(lambda x: x * 2)
print("\nDoubled DataFrame:\n", df10_doubled)

print("\n" + "=" * 60)
print("11. Lambda Function: New Column as Sum of Existing Columns (6 rows, 3 cols)")
print("=" * 60)
df11 = pd.DataFrame(np.random.randint(1, 20, size=(6, 3)), columns=['Val1', 'Val2', 'Val3'])
df11['Total_Sum'] = df11.apply(lambda row: row['Val1'] + row['Val2'] + row['Val3'], axis=1)
print(df11)

print("\n" + "=" * 60)
print("12. Text Data: Convert Series Strings to Uppercase")
print("=" * 60)
series12 = pd.Series(['python', 'pandas', 'machine learning', 'artificial intelligence', 'data science'])
print("Original Series:\n", series12)
series12_upper = series12.str.upper()
print("\nUppercase Series:\n", series12_upper)

print("\n" + "=" * 60)
print("13. Text Data: Extract First 3 Characters")
print("=" * 60)
series13 = pd.Series(['Mangaluru', 'Karnataka', 'Sahyadri', 'Engineering', 'Robotics'])
print("Original Series:\n", series13)
series13_slice = series13.str[:3]
print("\nFirst 3 Characters Extracted:\n", series13_slice)
```
